1. Arithmetic Mean Method

```python
n = int(input("Enter the number of machines: "))

machines = {}

for i in range(n):
    name = input(f"\nEnter machine {i + 1} name: ")

    rewards = list(map(int, input(
        f"Enter rewards for Machine {name} separated by space: "
    ).split()))

    machines[name] = rewards


q_values = {}

print("\n--- Estimated Q Values ---")

for machine, rewards in machines.items():

    q = sum(rewards) / len(rewards)
    q_values[machine] = q

    print(f"\nMachine {machine}")
    print("Rewards:", rewards)
    print(f"Q({machine}) = {q}")


# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", selected_machine)
```

2. Greedy Multi-Armed Bandit using Incremental Update


```python
n = int(input("Enter the number of machines: "))

# Dictionaries to store Q values and number of plays
Q = {}
N = {}

for i in range(n):

    machine = input(f"\nEnter machine {i + 1} name: ")

    rewards = list(map(int, input(
        f"Enter rewards for Machine {machine} separated by space: "
    ).split()))

    # Initial values
    Q[machine] = 0
    N[machine] = 0

    print(f"\nMachine {machine}")

    # Update Q after every reward
    for reward in rewards:

        # Increment number of plays
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (
            reward - Q[machine]
        ) / N[machine]

        print(f"Reward = {reward}")
        print(f"N({machine}) = {N[machine]}")
        print(f"Q({machine}) = {Q[machine]:.2f}")

    print()


# Display final estimated values
print("--- Final Q Values ---")

for machine in Q:
    print(f"Q({machine}) = {Q[machine]:.2f}")


# Greedy algorithm selects highest Q value
selected_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", selected_machine)
```
